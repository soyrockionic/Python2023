```python
import pandas as pd
#df = pd.read_csv('productos.csv', sep='|')
```

```python
df = pd.read_csv('productos.csv', sep='|' , usecols=['productos_descripcion', 'productos_precio_lista'])
print(df.head(10))
```

```python
print(df[df['productos_precio_lista'] >= 2500.0].head())
```

```python
# Estadisticas adicionales utiles
print(f"\nTotal de productos: {len(df)}")
productos_caros = df[df['productos_precio_lista'] >= 2500.0]
print(f"Productos >= 2500: {len(productos_caros)}")
print(f"Precio promedio: {df['productos_precio_lista'].mean():.2f}")
```

```python
# 1-Total de productos
print(f"Total de productos: {len(df)}")

# 2-Precio promedio, minimo y maximo
print(f"\nPrecio promedio: {df['productos_precio_lista'].mean():,.2f}")
print(f"Precio maximo: {df['productos_precio_lista'].max()}")
print(f"Precio minimo: {df['productos_precio_lista'].min()}")

# 3-Los 5 productos mas caros
print(f"\nLista de los 5 productos mas caros:")
df_unicos = df.drop_duplicates(subset=['productos_descripcion'])
print(df_unicos.sort_values(by='productos_precio_lista', ascending=False).head())

# 4-Cuantos productos cuestan mas de $3000
print(f"\nCant productos que cuestan mas de 3000: {len(df[df['productos_precio_lista'] > 3000])}")

```

```python
# Buscar productos que contengan una palabra, sin importar si esta en mayuscula o minuscula
busqueda = df[df['productos_descripcion'].str.contains('alfajor', case=False, na=False)
              & (df['productos_precio_lista'] < 2000)]
print(busqueda.head(10))
```
