```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('ModalidadVirtual.csv', sep=',', usecols=['time', 'carrera', 'edad', 'sexo', 'trabajo'])

print(f"Mayores de 40:\n{df[df['edad'] > 40]}")

# Exportar a Excel
# df.to_excel('ModalidadVirtual.xlsx', index=False)
```

```python
# Promedio de edad de estudiantes mayores de 40
print(f"{df[df['edad'] > 40]['edad'].mean():.2f}")

# Promedio de edad de los que estudian derecho
print(f"{df[df['carrera'] == 'Derecho']['edad'].mean():.2f}")
```

```python
# Cuantos estudiantes hay por estado de trabajo?
print("\nEstudiantes por situacion laboral:")
print(df['trabajo'].value_counts().to_string())

# Cuantos hombres vs mujeres hay en total?
print("\nEstudiantes por sexo:")
print(df['sexo'].value_counts().to_string())
```

```python
count = df['trabajo'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(count, labels=count.index, autopct='%1.1f%%', colors=plt.cm.Set3.colors)
plt.title('Distribucion de Estudiantes por Situacion Laboral')
plt.show()
```

```python
# Contar las frecuencias primero
count = df['trabajo'].value_counts()

count.plot(kind='bar', colormap='Set3')
plt.title('Estudiantes por Situacion Laboral')
plt.ylabel('Cantidad')
plt.xlabel('\nTrabajo')
plt.show()
```

```python
# Promedio edad por carrera
print(df.groupby('carrera')['edad'].mean().round(2).to_string())
```
